```python
import pandas as pd
import numpy as np
from pathlib import Path

folder = Path(r"C:\Users\ADMIN\ADAS_Program8")
folder.mkdir(parents=True, exist_ok=True)

rng = np.random.default_rng(42)
rows = []

for label in ["Alert", "Drowsy"]:
    for i in range(60):
        if label == "Alert":
            ratio = rng.uniform(0.05, 0.25)
            duration = rng.uniform(100, 350)
            blink = rng.integers(12, 25)
        else:
            ratio = rng.uniform(0.40, 0.85)
            duration = rng.uniform(600, 1500)
            blink = rng.integers(2, 10)

        rows.append([ratio, duration, blink, label])

data = pd.DataFrame(rows, columns=[
    "closure_ratio",
    "avg_closure_ms",
    "blink_rate",
    "label"
])

data.to_csv(folder / "eye_closure.csv", index=False)

print("Dataset created successfully!")
print("Total records:", len(data))
print(data.head())
```
